```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
```

```python
dir = Path("/Users/wm722/Documents/SCDs/training/fold_Brazil")
```

```python
populations = np.load(dir / "res_size_distributions.npz")

res_df = pd.Series(populations['seen_resolutions'], name='Resolution (m)')
pixel_size_df = pd.Series(populations['object_sizes'], name='Object Size (pixels)') 
diameter_df = pd.Series(populations['real_object_diameters'], name='Real Object Diameter (m)') 
```

```python
res_df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for ax, df in zip(axes, [res_df, pixel_size_df, diameter_df]):
    df.plot.hist(ax=ax, bins=50, alpha=0.7)
    ax.set_title(df.name)
    ax.set_xlabel(df.name)
    ax.set_ylabel('Frequency')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

data = np.load('/Users/wm722/Documents/SCDs/results/Veto/UK/veto_embeddings.npz')

embeddings = data["embeddings"]
labels = data["labels"].astype(int)

pca = PCA(n_components=2)
components = pca.fit_transform(embeddings)

plt.figure(figsize=(8, 6))

for label, name, colour in [
    (0, "Reject", "tab:red"),
    (1, "Retain", "tab:blue"),
]:
    mask = labels == label
    plt.scatter(
        components[mask, 0],
        components[mask, 1],
        label=name,
        color=colour,
        alpha=0.6,
        s=20,
    )

plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]:.1%})")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]:.1%})")
plt.title("Veto classifier embeddings - PCA")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

data = np.load('/Users/wm722/Documents/SCDs/results/Veto/UK/veto_embeddings.npz')
X = data["embeddings"]
y = data["labels"].astype(int)

norms = np.linalg.norm(X, axis=1)
keep = norms < np.percentile(norms, 99.5)

Xs = StandardScaler().fit_transform(X[keep])
Z = PCA(n_components=2).fit_transform(Xs)
yk = y[keep]

fig, ax = plt.subplots(figsize=(8, 6))
for lab, name, col in [(0, "Reject", "tab:red"), (1, "Retain", "tab:blue")]:
    m = yk == lab
    ax.scatter(Z[m, 0], Z[m, 1], s=6, alpha=0.3, c=col, label=name)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")

ax.set_title("Veto classifier embeddings (PCA, top 0.5% by norm dropped)")
ax.legend(markerscale=3)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import numpy as np
from pathlib import Path
import sys
sys.path.append("../../helpers/")
import geostats as gs
```

```python
STATS_PATH = Path('../../../stats/')
RES = 1
region_names = ['weymouth_only', 'wells_only', 'wellington_only']
```

```python
analyser = gs.StatsStack(region_names, STATS_PATH)
```

```python
analyser.plot_metrics()
```

```python
# average metrics for Quantock region

stats = analyser.all_region_metrics['wellington_only']
area = stats['areas'].mean()
depth = stats['depths'].mean()
mal = stats['major_axis_lengths'].mean()

print(f'Area: {area}\ndepth: {depth}\nmal: {mal}')
```

```python
best_fit_circle_r = (area / np.pi) ** 0.5

print(best_fit_circle_r)
```

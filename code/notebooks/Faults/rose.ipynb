```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

strikes = np.asarray([20, 175, 35, 50, 175, 55, 10, 45, 60, 40, 0, 5, 5, 5, 5, 10, 10, 20, 0, 0, 5, 5, 5, 10, 10, 5, 5, 130, 130, 130, 0, 0, 
           150, 150, 170, 0, 0, 175, 70, 70, 20, 10, 70, 70, 40]) + 90

bin_width = 10

bins = np.arange(0, 181, bin_width)
counts, edges = np.histogram(strikes, bins=bins)

centres = np.deg2rad((edges[:-1] + edges[1:]) / 2)
centres_mirror = centres + np.pi
width = np.deg2rad(bin_width)
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection="polar")

# draw both halves
for c, r in zip(centres, counts):
    ax.bar(c, r, width=width, bottom=0, color='blue')
    ax.bar(c + np.pi, r, width=width, bottom=0, color='blue')

ax.set_theta_zero_location("N") 
ax.set_theta_direction(-1)        

max_r = counts.max()
ax.set_ylim(0, max_r * 1.15)
ax.yaxis.set_visible(False)

# compass labels
ax.set_xticks(np.deg2rad([0, 45, 90, 135, 180, 225, 270, 315]))
ax.set_xticklabels(["N", "NE", "E", "SE", "S", "SW", "W", "NW"],
                   color="#cccccc", fontsize=10, fontfamily="monospace")

ax.tick_params(axis="x", pad=8)
ax.spines["polar"].set_color("#333333")
ax.grid(color="#333333", linewidth=0.5, linestyle="--", alpha=0.5)



plt.tight_layout()
plt.show()
```

```python
import rasterio as rio
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive/')
from scipy.ndimage import median_filter, gaussian_filter
import numpy as np
```

```python
with rio.open('/content/drive/MyDrive/IRP/Data/USA/ClippedMergedBarnhart.tif') as src:
    dem = src.read(1, masked=True).astype(np.float32)
    dem = dem.filled(np.nan)
    dem = dem[10000:17500, 10000:20000]
```

```python
smoothed = gaussian_filter(dem, sigma=30)
```

```python
plt.imshow(dem)
```

```python
plt.imshow(smoothed)
```

```python
plt.imshow(dem-smoothed)
```

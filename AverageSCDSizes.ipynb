```python
from google.colab import drive

drive.mount('/content/drive/')

import numpy as np
from glob import glob
import sys
sys.path.append('/content/drive/MyDrive/IRP/scripts')

from ModelCompare import REGION_GROUPS, _region_paths
```

```python
stats = {}

for region in REGION_GROUPS.keys():
  print(f'Evaluating {region}')
  paths = _region_paths(region)
  to_check = len(paths)
  instances = []

  for i, p in enumerate(paths):

    d = np.load(p)

    try:
      len(d['object_ids'])
    except:
      continue

    if i % 10 == 0:
      print(f'Checking image {i} / {to_check}')

    image_area = 512 * 512 * (d['res'] ** 2)
    scd_area = image_area * d['scd_pixel_fraction']
    scd_count = len(d['object_ids'])

    img_avg_area = scd_area / scd_count
    r_avg = (img_avg_area / np.pi) ** 0.5
    instances.append(r_avg)

  stats[region] = instances
```

```python
import pickle

with open('/content/drive/MyDrive/IRP/results/scd_sizes.pkl', 'wb') as f:
  pickle.dump(stats, f)
```

```python
import rasterio as rio
from google.colab import drive
drive.mount('/content/drive/')
from scipy.ndimage import median_filter
```

```python
raw_dem_path = '/content/drive/MyDrive/IRP/Data/UKQuantock/dem_wellington_only.tif'
out_dem_path = '/content/drive/MyDrive/IRP/Data/UKQuantock/quan_extracted_DTM.tif'
```

```python
with rio.open(raw_dem_path) as src:
  dsm = src.read(1)
  profile = src.profile
```

```python
dtm = median_filter(dsm, size=5)

profile.update(dtype='float32', count=1, compress='lzw')
with rio.open(out_dem_path, 'w', **profile) as dst:
    dst.write(dtm.astype('float32'), 1)
```
